```python
import pandas as pd
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
data = {
    "X": [ 1,2,2,8,9,25],
    "y": [2,2,3,8,8,30]
}
```

```python
df= pd.DataFrame(data)
```

```python
db = DBSCAN (eps= 3, min_samples= 2)
```

```python
df["Cluster"] = db.fit_predict(df[["X","y"]])
```

```python
print(df)
```

```python
plt.scatter(df["X"], df["y"], c = df["Cluster"], cmap = 'rainbow', s = 100)
plt.show()
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
neighbors = NearestNeighbors(n_neighbors=2)
neighbors_fit = neighbors.fit(df)
```

```python
distance,indices = neighbors_fit.kneighbors(df)
```

```python
distances = sorted (distance[:,1])
```

```python
distances
```

```python
plt.plot(distances)
```

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
```

```python
sales = [120,130,128,140,145,150,155,160,165,170]
```

```python
df = pd.DataFrame({"Sales": sales})
```

```python
model = ARIMA(df["Sales"],order = (1,1,1))
```

```python
model_fit = model.fit()
```

```python
forecast = model_fit.forecast(steps = 5)
```

```python
print(forecast)
```

```python
plt.plot(df["Sales"])
plt.plot(range(len(df),len(df)+5),forecast,label = "Forecast")
plt.show()
```
