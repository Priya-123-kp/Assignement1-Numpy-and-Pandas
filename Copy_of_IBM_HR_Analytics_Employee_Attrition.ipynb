4.3 Outlier Detection and Treatment

````python
plt.figure(figsize=(8,5))
sns.boxplot(df['MonthlyIncome'])
plt.title('Box Plot of Monthly Income')
plt.show()
````

IQR Method

````python
Q1 = df['MonthlyIncome'].quantile(0.25)
Q3 = df['MonthlyIncome'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
UPPER_bound = Q3 + 1.5 * IQR
df = df[(df['MonthlyIncome'] >= lower_bound) & (df['MonthlyIncome'] <= UPPER_bound)]
````

````python
plt.figure(figsize=(8,5))
sns.boxplot(df['MonthlyIncome'])
plt.title('Box Plot of Monthly Income')
plt.show()
````

Z-Score Method

````python
from scipy import stats
z_scores = np.abs(stats.zscore(df['MonthlyIncome']))
df = df[z_scores < 3]
````

````python
df
````

````python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['MonthlyIncome'], color='lightgreen')
plt.title('Monthly Income (After Z-score Cleaning)')
plt.show()
````


Observation:

The IQR method was used to identify and remove extreme values.

Time Series Projects

````python

from sklearn.metrics import davies_bouldin_score

db_score = davies_bouldin_score(X, cluster_labels)
print("Davies-Bouldin Score:", db_score)

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, pred_rf)

rmse = np.sqrt(mean_squared_error(y_test, pred_rf))

non_zero_indices = y_test != 0
if non_zero_indices.any():
    mape = np.mean(np.abs((y_test[non_zero_indices] - pred_rf[non_zero_indices]) / y_test[non_zero_indices])) * 100
else:
    mape = np.nan

print("MAE:", mae)
print("RMSE:", rmse)
print("MAPE:", mape)
````

MAE (Mean Absolute Error)

Average absolute difference between actual and predicted values.

RMSE (Root Mean Squared Error)

Gives higher penalty to large prediction errors.

MAPE (Mean Absolute Percentage Error)

Measures prediction error as a percentage.
