Task 1- Linear Algebra Implementation

```python
import numpy as np
```

```python
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
```

```python
matrix_a = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])

matrix_b = np.array([
    [1, 2, 0],
    [3, 1, 4],
    [2, 5, 1]
])
```

```python
print("Vector A:", vector_a)
print("Vector B:", vector_b)
```

```python
print("Vector Addition")
print(vector_a + vector_b)
```

```python
print("Vector Subtraction")
print(vector_a - vector_b)
```

```python
print("Dot Product")
print(np.dot(vector_a, vector_b))
```

```python
print("Matrix Addition")
print(matrix_a + matrix_b)
```

```python
print("Matrix Multiplication")
print(np.matmul(matrix_a, matrix_b))
```

```python
det = np.linalg.det(matrix_a)
print("Determinant of Matrix A")
print(det)
```

```python
if det != 0:
    print("Inverse of Matrix A")
    print(np.linalg.inv(matrix_a))
```

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)
```

```python
print("Eigenvalues")
print(eigenvalues)
```

```python
print("Eigenvectors")
print(eigenvectors)
```

```python
customer_segments = ["Premium", "Regular"]
```

```python
segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}
```

```python
print("Conditional Probability")
for segment, prob in segment_purchase_probability.items():
    print(f"P(Purchase | {segment}) = {prob}")
```

```python
P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_A) / P_B
```

```python
print("Bayes Theorem Result")
print("P(A|B) =", P_A_given_B)
```

```python
n_trials = 20
success_prob = 0.65
```

```python
uniform_low = 0
uniform_high = 1

uniform_distribution = np.random.uniform(
    low=uniform_low,
    high=uniform_high,
    size=1000
)
```

Task 3 – Descriptive Statistics

```python
import numpy as np
import pandas as pd
from statistics import mode
from scipy.stats import skew, kurtosis
```

```python
sales_data = [
    120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
    190, 200, 210, 195, 185, 175, 165, 155, 145, 135
]
```

```python
mean_value = np.mean(sales_data)
```

```python
median_value = np.median(sales_data)
```

```python
mode_value = mode(sales_data)
```

```python
range_value = max(sales_data) - min(sales_data)
```

```python
variance_value = np.var(sales_data)
```

```python
std_value = np.std(sales_data)
```

```python
skewness_value = skew(sales_data)
```

```python
kurtosis_value = kurtosis(sales_data)
```

```python
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
print("Range:", range_value)
print("Variance:", variance_value)
print("Standard Deviation:", std_value)
```

```python
print("Skewness:", skewness_value)
print("Kurtosis:", kurtosis_value)
```

```python
if skewness_value > 0:
    interpretation = "Positively Skewed"
elif skewness_value < 0:
    interpretation = "Negatively Skewed"
else:
    interpretation = "Symmetric"
print("Distribution Interpretation")
print(interpretation)
```
