```python
import pandas as pd
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print(marks)
```

```python
print("1st Rank Mark:")
print(marks[0])
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
print("3rd Rank Mark:")
print(marks.iloc[2])
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
marks['Rank1'] = 100
print("After Updating Rank1 Mark:")
print(marks)
```

```python
marks = marks.drop('Rank5')
print("\nAfter Removing Last Rank:")
print(marks)
```

```python
cgpa = marks / 10
print("\nCGPA:")
print(cgpa)
```

# Creating the DataFrame

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
```

```python
print(transactions)
```

```python
print("Transaction Data frame:")
print(transactions)
```
