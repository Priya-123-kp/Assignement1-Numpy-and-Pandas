```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Line Chart- Fare Over Time

```python
df = sns.load_dataset("taxis")
```

```python
df['pickup'] = pd.to_datetime(df['pickup'])
```

```python
df=df.sort_values('pickup')
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['pickup'],df['fare'])
plt.title("Fare Over Time")
plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.show()
```

Bar Chart

```python
borough_fare = df.groupby('pickup_borough')['fare'].sum()
```

```python
borough_fare.plot(kind='bar',figsize =(8,5))

plt.title("Totla Fare by Borough")
plt.xlabel("pickup Borough")
plt.ylabel("Total Fare")
plt.show()
```

Pie Chart

```python
payment_counts = df['payment'].value_counts()
payment_counts.plot(kind='pie',figsize=(6,6),autopct='%1.1f%%')
plt.title('Payment Method Distribution')
plt.ylabel("")
plt.show()
```

Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df['distance'],bins=20)
plt.title("Distribution of Distance")
plt.xlabel('Distance')
plt.ylabel('Frequency')
plt.show()
```

Box Plot

```python
plt.figure(figsize =(10,6))
sns.boxplot(x='payment',y='tip',data=df)
plt.title("Tip Distribution by Pickup Borough")
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.show()
```

Visualizations using Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset("taxis")
```

Count Plot

```python
plt.figure(figsize=(8,5))
sns.countplot(x='pickup_borough',data=df)
plt.title("Number of Trips by Pickup Borough")
plt.xlabel('Pickup Borough')
plt.ylabel('Trip Count')
plt.show()
```

Scatter Plot

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x ='distance', y ='fare',hue ='pickup_borough',data=df)
plt.title("Fare vs Distance ")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.show()
```

Heatmap

```python
num_cols = df[['distance', 'fare','tip','tolls','total']]
corr_matrix = num_cols.corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix,annot=True,cmap='coolwarm',fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

Pair Plot

```python
sns.pairplot(df[['distance','fare','tip','total','pickup_zone']],hue='pickup_zone')
plt.show()
```

Violin Plot

```python
plt.figure(figsize=(10,6))
sns.violinplot(x='payment',y='fare',data=df)
plt.title("Fare Distribution by Payment Method")
plt.xlabel('Payment Method')
plt.ylabel('Fare')
plt.show()
```
